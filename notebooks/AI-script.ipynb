```python
import requests
import pandas as pd
import os
from datetime import datetime
from io import StringIO
from datetime import datetime, timezone

# --------------------------------------------------
# NASA FIRMS API Einstellungen
# --------------------------------------------------

MAP_KEY = "01b9b20f1c9e80d44560acd21f5a79a7"

# Bounding Box für Südamerika
# west,south,east,north
AREA = "-85,-57,-32,14"

# --------------------------------------------------
# Sensoren (3 Stück)
# --------------------------------------------------

SENSORS = [
    "VIIRS_NOAA21_NRT",
    "VIIRS_NOAA20_NRT",
    "VIIRS_SNPP_NRT"
]

# Letzte 24h = day range 1
DAY_RANGE = 1

# --------------------------------------------------
# Output Ordner (existiert bereits)
# --------------------------------------------------

OUTPUT_DIR = os.path.join("..", "data", "raw")
os.makedirs(OUTPUT_DIR, exist_ok=True)

print("Starte Download für alle Sensoren...")

for SOURCE in SENSORS:

    print(f"Lade Daten für Sensor: {SOURCE}")

    url = (
        f"https://firms.modaps.eosdis.nasa.gov/api/area/csv/"
        f"{MAP_KEY}/{SOURCE}/{AREA}/{DAY_RANGE}"
    )

    print(url)

    try:
        response = requests.get(url)
        response.raise_for_status()

        df = pd.read_csv(StringIO(response.text))

        print(f"Datensätze gefunden ({SOURCE}): {len(df)}")

        timestamp = datetime.utcnow().strftime('%Y%m%d_%H%M%S')

        filename = f"firms_{SOURCE}_south_america_{timestamp}.csv"
        file_path = os.path.join(OUTPUT_DIR, filename)

        if not df.empty:
            df.to_csv(file_path, index=False)
            print(f"Gespeichert: {file_path}")
        else:
            print("Keine Daten für diesen Sensor.")

    except Exception as e:
        print(f"Fehler bei {SOURCE}: {e}")

print("Fertig.")
```
